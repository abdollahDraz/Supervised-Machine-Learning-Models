```python
# Voting
voting_hard = VotingClassifier(estimators= [("log",log),("dt",dt),("svm",svm)] , voting= "hard")
voting_soft = VotingClassifier(estimators= [("log",log),("dt",dt),("svm",svm)] , voting= "soft")
# training
voting_hard.fit(X_train,y_train)
voting_soft.fit(X_train,y_train)
# Predection
voting_hard_pred = voting_hard.predict(X_test)
voting_soft_pred = voting_soft.predict(X_test)
# Evaluation
from sklearn.metrics import accuracy_score
print("The Accuracy  of voting hard is :- " , accuracy_score(voting_hard_pred,y_test) )
print("The Accuracy of voting soft  is :- " , accuracy_score(voting_soft_pred,y_test))



```
